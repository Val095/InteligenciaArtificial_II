_El espejismo de la "exactitud" (accuracy)_

El escenario: nuestro clacificador acierta el 90% de las veces. Superficialmente, parece un modelo perfecto y listo para entrar en fase de producción.

La realidad: solo el 10% de las imágenes  con el número "5". Si programamos un modelo inútil que siempre, ciegamente diga "no es un 5", también tendrá un 90% de exactitud.

La exatitud (Accuracy) es una métrica profundamente engañosa cuando se trabaja con datos desequilibrados (skewed datasets), donde un a clase es mucho más frecuente que la otra.

_La matriz de confusión_

Verdaderos negativos **(TN)**: Dijo que NO era 5, y acertó.

Falsos positivos **(FP)**: Dijo que ERA 5, pero era otro número. Error tipo 1

Falsos Negativos **(FN)**: Dijo que NO era 5, pero si era. Erro tipo 2

Verdaderos positivos **(TP)**: Dijo que ERA 5 y acertó.

La matriz de confusión nos permite ver exactamente donde se equivoca el modelo, no solo cuánto. Separa las falsas alarmas de las amenazas críticas ignoradas.

_Precisión vs Exhaustividad_

Precisión (Calidad): ¿Qué tan exacto eres cuando dices "sí"?

Fórmula: TP / (TP + FP)

Ejemplo de negocio: Un filtro de videos infantiles. Es imperativo rechazar videos inofensivos (alta precisión) antes de dejar pasar un solo contenido inapropiado.

Exhaustividad/Recall (Cantidad): ¿Qué porcentaje de los casos reales lograste atrapar?

Fórmula: TP / (TP+FN)

Ejemplo de negocio: Cámaras de seguridad en un aeropuerto. Es preferible soportar múltiples falsas alarmas (alta exhaustividad) que dejar escapar a un verdadero intruso.

F1 Score: La media armónica. Solo alcanza niveles alos si la balanza entre Precisión y Exhaustividad está equilibrada.

_El umbral de decisión: El equilibrio (Trade-Off)_

**Umbral bajo (mover a la izquierda)**

-El modelo se vuelve hiper-sensible y atrapa todo.

-Aumenta la exhaustividad (Recall).

-Consecuencia: Cae la precisión. Se genera exceso de falsas alarmas y costos operativos innecesarios.

**Umbral alto (Mover a la derecha):**

-El modelo solo predice "sí" cuando estpa matemáticamente seguro.

- Aumenta la precisión.

-Consecuencia: Cae la exhaustividad. Se escapan verdaderos positivos (oportunidades o amenazas reales).
