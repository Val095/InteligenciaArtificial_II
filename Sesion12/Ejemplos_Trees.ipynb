**RANDOM FOREST - EJEMPLO DE LA VIDA REAL**

Decidir si comprar una casa

🌳 Árbol 1: analiza área, habitaciones y ubicación.

🌳 Árbol 2: analiza antigüedad, baños y garaje.

🌳 Árbol 3: analiza área, estrato y distancia al centro.

🌳 Árbol 4: analiza habitaciones, baños y estado de la vivienda.

🌳 Árbol 5: analiza ubicación, área y antigüedad.

Cada árbol hace su propia predicción:

Árbol 1 → $300.000

Árbol 2 → $320.000

Árbol 3 → $310.000

Árbol 4 → $305.000

Árbol 5 → $315.000

El Random Forest combina las predicciones para obtener una respuesta más estable, por ejemplo:

Precio estimado → $310.000
