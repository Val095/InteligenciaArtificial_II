1. Valor del ML

2. TIPOS DE SISTEMAS

El aprendizaje supersivado el algoritmo necesita un maestro

En el aprendizaje no supervisado obliga a explorar al algoritmo a ciegas

Aprendizaje por refuerzo el agente informatico toma decisiones, obteniendo recompensas a punta de prueba y error.

3. FLUJO DE DATOS

Aprendizaje por lotes - rigido, poder computacional enorme.

Aprendizaje en líneas - permite una adaptación continua a nuevos datos.

Aprendizaje fuera del núcleo - es una técnica de Machine Learning diseñada para manejar conjuntos de datos que son demasiado grandes para caber en la memoria RAM de la computadora.

4. estrategias de generalización

Aprendizaje basado en instancias- es un paradigma de Machine Learning donde el modelo no genera una función objetivo abstracta o explícita durante la fase de entrenamiento. En su lugar, almacena los datos de entrenamiento en la memoria y pospone la generalización hasta que se recibe una nueva consulta (query).

Aprendizaje basado en modelos - es el paradigma opuesto al aprendizaje basado en instancias. En lugar de memorizar los datos, el algoritmo utiliza el conjunto de entrenamiento para construir una función matemática o regla general (un "modelo") que capture los patrones subyacentes.

5. Desafíos principales

La salud del sistema de predicción depende de la limpieza de los datos, para extraer las variables que de verdad importan.

El sobreajuste ocurre si un modelo complejo memoriza el ruido

E Infrajuste es cuando el modelo es demasiado bajo
