Aprendizaje supervisado, no supervisado y por refuerzo

Aprendizaje supervisado - EL entrenamiento de los modelos depende de datos etiquetados previamente por humanos. La tarea principal es la clasificación, la segunda la regresion. Arboles de decision - preguntas l+ogicas

Aprendizaje no supervisado- Se enfrenta a un oceano de datos crudos con el fin de encontrar el orden por su propia cuenta.

Revela la estructura intrínseca de los datos libre de los sesgos que pueden proyectar los humanos

Por refuerzo - Entorno dinámico en tiempo real. Retroalimentación en tiempo real, se recibe una recompensa. Permite que se hagan tareas complejas en robótico. Planes de accion secuencial, priorizando el resultado final en entornos de alta incertidumbre.

La computacion ha superado su rol original al identificar patrones estructurales por su cuenta, ahora funciona como un sistema de predicción.
