4. Memoria a corto plazo

5. Celdas de memoria

LSTM - Memoria a corto y largo plazo, se divide en dos carriles, corto y largo plazo, permite que la informacion viaje intacta a través del tiempo.
Está controlado por tres compuertas lógicas: Forget gate, Input gate, Outout gate

GRU - Unidad recurrente con compuerta, tiene un diseño más limpio en donde combina los dos carriles y simplifica las compuertas en una sola, consume menos recursos

6. Secuencias Largas

Hacer preprocesamiento usando capas convolucionales en una sola dimension. Especial para que no se abrumen

WaveNet - Aplica capas convolucionales, procesa miles pasos de tiempo de manera paralela. Rompe las barrera de como se comprende el tiempo en los datos

tasa de dilatación, la primera capa analiza datos pegados unos a otros, la siguiente de uno de por medio y así suscesivamente, cada capa duplica la forma en la que analizan los datos, las altas entienden a largo plazo, mientras que las capas bajas revisan todo milimetricamente
